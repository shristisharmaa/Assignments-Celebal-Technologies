```python
spark.sql("""
SELECT COUNT(*)
FROM retail_gold.fact_orders VERSION AS OF 0
""").show()
```

```python
spark.sql("""
SELECT
COUNT(*) AS total_orders,
COUNT(DISTINCT order_id) AS unique_orders
FROM retail_gold.fact_orders
""").show()
```

```python
spark.sql("""
SELECT
SUM(CASE WHEN customer_sk IS NULL THEN 1 ELSE 0 END) AS null_customer_sk,
SUM(CASE WHEN product_sk IS NULL THEN 1 ELSE 0 END) AS null_product_sk,
SUM(CASE WHEN store_sk IS NULL THEN 1 ELSE 0 END) AS null_store_sk
FROM retail_gold.fact_orders
""").show()
```
