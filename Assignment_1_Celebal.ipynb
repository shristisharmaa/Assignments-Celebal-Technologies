3. Handle missing values (identify, fill/drop).

```python
# MISSING VALUES BEFORE CLEANING

df.isnull().sum()
```

```python
# Fill missing values in discount column

df["discount"].fillna(df["discount"].mean(), inplace=True)
```

```python
# MISSING VALUES AFTER CLEANING

df.dropna(inplace=True)

print("Missing values after cleaning:")
print(df.isnull().sum())
```

4. Perform basic operations (filter rows, select columns).

```python
# Filter rows with rating greater than 4

filtered_data = df[df["rating"] > 4]

print(filtered_data[["title", "rating", "final_price", "category"]].head(10))
```

```python
# Filter products with discount greater than 50

print(df[df["discount"] > 50][["title", "discount"]].head())
```

```python
# Filter products from sweaters category

print(df[df["category"] == "sweaters"][["title", "category"]].head())
```

```python
# Select specific columns

print(df[["title", "rating", "final_price"]].head())
```

5. Remove duplicates.

```python
# Check duplicate rows
print(df.duplicated().sum())

# Remove duplicates
df.drop_duplicates(inplace=True)
print(df.duplicated().sum())
```

6. Create a derived column (total_amount = price * quantity).

```python
# Create quantity column

df["quantity"] = 1

# Create derived column

df["total_amount"] = df["initial_price"] * df["quantity"]

print(df[["initial_price", "quantity", "total_amount"]].head())
```

7. Save the cleaned dataset as a new CSV file.

```python
# Save cleaned dataset as a new CSV file

df.to_csv("cleaned_dataset.csv", index=False)

print("Cleaned dataset saved successfully")
```
